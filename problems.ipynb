```python
#2 the game() function in a program lets a user play a game and returns the score as a integer.
# You need to read a file'Hi-score.txt' which is either blank or contains the previous Hi score.
#

import random
# r = random.randint(1,100)
# print(r) 
def game():
    print("You are playing a game")
    score = random.randint(1,100)
    with open("hiscore.txt") as f:
         hiscore = f.read()
         if(hiscore != ""):
            hiscore = int(hiscore)
         else:
            hiscore = 0
    print(f"Your score is:{score}")
    if(score>hiscore):
        with open("hiscore.txt","w") as f:
            f.write(str(score))
    return score
game()
```

```python
#3 write a program to generate multiplication tables from 2 to 20 and write it to the different files.
# place these files in a folder for a 13 year old.

import os

def generateTable(n):
    table = ""
    for i in range(1, 11):
        table += f"{n} X {i} = {n*i}\n"
    
    # Create folder if not exists
    os.makedirs("DAY9/table", exist_ok=True)

    # Write to file
    with open(f"DAY9/table/table{n}.txt", "w") as f:
        f.write(table)

# Generate tables from 1 to 20
for i in range(1, 21):
    generateTable(i)
```
