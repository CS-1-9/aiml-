```python
def fact(num):
    result = 1 
    while num>1:
        result = result*num
        num = num - 1
    return result
print(fact(int(input("Enter the number:"))))
```

```python
def fact(num):
    Result = 1
    for i in range(1,num+1):
        Result = Result*i
    return Result
print(fact(7))
```

```python
# fibonacci series
def fibonacci(n):
    fib_sequence = [0,1]
    while len(fib_sequence)<n:
        fib_sequence.append(fib_sequence[-1]+fib_sequence[-2])
    return fib_sequence
print(fibonacci(5))
```

```python
# recursion
# function will call to itself
def factorial(n):
    if n==0 or n==-1:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(5))
print(factorial(1))
print(factorial(0))
```

```python
# using recursion fibonacci 
def fibonacci_recursion(n):
    if n<=0:
        return []
    
    elif n==1:
        return [0]
    elif n==2:
        return [0,1]
    else:
        fib_sequence = fibonacci_recursion(n-1)
        fib_sequence.append(fib_sequence[-1]+fib_sequence[-2])
    return fib_sequence
fibonacci_recursion(5)
```
