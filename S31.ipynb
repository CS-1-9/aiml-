```python
# Slicing example
text = "Python Programming"

print(text)           # Prints the whole string

# start:stop:step
print(text[0:5])      # Prints characters from index 0 to 4 → 'Pytho'
print(text[::])       # Prints the entire string → 'Python Programming'
print(text[::-1])     # Reverses the string → 'gnimmargorP nohtyP'
print(text[-40:-1])   # Slices safely even if index out of range → 'Python Programmin'
```
