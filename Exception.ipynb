```python
a = int(input("Enter a value"))
b = int(input("Enter b value "))
c = a/b
print(c)
```

```python
try:
    print(10/0)
except ZeroDivisionError:
    print(10/2)
    print("Not divisible by zero")
```

```python
try:
    a = int(input("Enter a value"))
    b = int(input("Enter b value"))
    c = a/b
    print("answer",c)
except Exception as e:
    print("Exception catch",e)

```

```python
# multiple except block
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    print(x/y)
except ZeroDivisionError:
    print("cant divide by zero")
except ValueError:
    print("please provide int value only")
```

```python
try:
    # we are not define variable j
    print(j)
except NameError:
    print("please define the variable")
# except :
else:
    print("Name error exception solved the problem, dont worry ")
```

```python
# finally block 
# the block which run compulsory if error or not 

try:
    # we are not define variable j
    print(j)
except NameError:
    print("please define the variable")
# except :
finally:
    print("Name error exception solved the problem, dont worry ")
```
