```python
# list of nodes 
f = open("demo2.txt","w")
list = ["bunny \n","chinu \n","monu \n"]
f.writelines(list) # multiple lines
f.close()
```

```python
# reading the files
# read() # total data from the file 
# read(n) > read 'n' character from the file 
# readlines()> read only one line 
# readlines()> to read all the lines into the list 
```

```python
file = open("demo1.txt","r")
filecontent = file.read()
print(filecontent)
```

```python
# read(n)
f = open("demo1.txt","r")
file_content = f.read(15)
print(file_content)
```

```python
# readline # to read only one line 
file = open("demo1.txt","r")
con = file.readline()
print(con)
print(con)
```

```python
# readlines # to read multiple lines
file = open("demo1.txt","r")
con  = file.readlines()
print(con)
file.close()
```

```python
# using the with keyword
with open("demo1.txt","r") as file:
    content = file.read()
    print(content)
# you no need to close the file when using the statement
```

```python
# append mode "a"
s = "we are in the acharya college"
file = open("demo1.txt","a")
file.write(s)
print("file updated")
file.close()
```

```python
# pickle file
import pickle
data = {"name":"john","age":30,"city":"bengaluru"}
with open("data.pkl","wb") as file:
    pickle.dump(data,file)
```

```python
with open("data.pkl","rb") as file:
    loaded_data =pickle.load(file)
    print(loaded_data)
```
