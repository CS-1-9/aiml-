```python
#range method 
# print "Happy Birthday" 10 times 

for x in range(10):
    print("Happy Birthday")
```

```python
# print from 1 to 100

for x in range(101):
    print(x)
```

```python
# display odd number from 0 to 20

for x in range(21):
    if (x%2 !=0):
        print(x)
```

```python
# display number from 10 to 1 in decending order

for x in range(10,0,-1):
    print(x)
```
